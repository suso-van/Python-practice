Problem 1 - Student Marks Filter


```python
def filter_student_marks(marks_dict, threshold):
    """
    Filters a dictionary of students and their marks, returning only those
    who scored greater than or equal to the specified threshold.
    """
    return {student: mark for student, mark in marks_dict.items() if mark >= threshold}

# Sample Student Data
student_marks = {
    "Alice": 85,
    "Bob": 58,
    "Charlie": 92,
    "David": 45,
    "Eva": 73
}

# Filter students with marks greater than or equal to 60
passing_students = filter_student_marks(student_marks, 60)
print("Students who passed (scored >= 60):")
for student, mark in passing_students.items():
    print(f"{student}: {mark}")
```

Below is the implementation of the `filter_student_marks` function, followed by a demonstration with sample data.

Problem 2- The patterns(Pattern V,X,N,W,Z


```python
def print_pattern_v(height=5):
    print("--- Pattern V ---")
    for i in range(height):
        for j in range(2 * height):
            if j == i or j == (2 * height - 2 - i):
                print("*", end="")
            else:
                print(" ", end="")
        print()

def print_pattern_x(height=5):
    print("\n--- Pattern X ---")
    if height % 2 == 0:
        height += 1
    for i in range(height):
        for j in range(height):
            if j == i or j == (height - 1 - i):
                print("*", end="")
            else:
                print(" ", end="")
        print()

def print_pattern_n(height=6):
    print("\n--- Pattern N ---")
    for i in range(height):
        for j in range(height):
            if j == 0 or j == height - 1 or i == j:
                print("*", end="")
            else:
                print(" ", end="")
        print()

def print_pattern_w(height=5):
    print("\n--- Pattern W ---")
    for i in range(height):
        for j in range(4 * height):
            # Outer left diagonal, inner diagonals (making the peak in the middle), outer right diagonal
            if j == i or j == (2 * height - 2 - i) or j == (2 * height + i) or j == (4 * height - 2 - i):
                print("*", end="")
            else:
                print(" ", end="")
        print()

def print_pattern_z(height=5):
    print("\n--- Pattern Z ---")
    for i in range(height):
        for j in range(height):
            if i == 0 or i == height - 1 or j == (height - 1 - i):
                print("*", end="")
            else:
                print(" ", end="")
        print()

# Run demonstrations
print_pattern_v()
print_pattern_x()
print_pattern_n()
print_pattern_w()
print_pattern_z()
```

Problem 3 -  Triangle ,diamond and hexa gon

```python
def print_triangle(height=5):
    print("--- Triangle Pattern ---")
    for i in range(height):
        # Print leading spaces
        print(" " * (height - i - 1), end="")
        # Print stars
        print("*" * (2 * i + 1))
    print()

def print_diamond(height=5):
    print("--- Diamond Pattern ---")
    # Upper half
    for i in range(height):
        print(" " * (height - i - 1), end="")
        print("*" * (2 * i + 1))
    # Lower half
    for i in range(height - 2, -1, -1):
        print(" " * (height - i - 1), end="")
        print("*" * (2 * i + 1))
    print()

def print_hexagon(side=4):
    print("--- Hexagon Pattern ---")
    # Top half (including middle)
    for i in range(side):
        print(" " * (side - i - 1), end="")
        print("*" * (side + 2 * i))
    # Bottom half
    for i in range(side - 2, -1, -1):
        print(" " * (side - i - 1), end="")
        print("*" * (side + 2 * i))
    print()

# Run demonstrations
print_triangle(5)
print_diamond(5)
print_hexagon(4)
```

Problem 5- Error or divisible by 5

```python
def check_divisibility_by_5(value):
    """
    Checks if a value is divisible by 5.
    Raises ValueError if the input is not a valid integer or float.
    """
    try:
        # Ensure numeric type conversion
        num = float(value)
        if num % 5 == 0:
            return f"{value} is divisible by 5."
        else:
            return f"{value} is NOT divisible by 5."
    except (ValueError, TypeError):
        raise ValueError(f"Error: '{value}' is not a valid number.")

# Test scenarios
test_cases = [25, 13, "50", "hello", 12.5, [1, 2]]

for case in test_cases:
    try:
        result = check_divisibility_by_5(case)
        print(result)
    except ValueError as e:
        print(e)
```
